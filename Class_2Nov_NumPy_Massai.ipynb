**Copy vs View**

```python
import numpy as np

a = np.array([10, 20, 30, 40])

b = a.copy()

b[0] = 999

print(a)
print(b)
```

```python
a = np.array([10, 20, 30, 40])

b = a.view()

b[0] = 999

print(a)
print(b)
```

```python
a = np.array([10, 20, 30, 40])
b = a[1:3]

print(b)
b[0] = 999

print(a)
print(b)
```

```python
a = np.array([10, 20, 30, 40, 50])

# Q1: Create a copy named b.
# Change b[1] to 999.
# Check whether a changes.

# Q2: Create a view named c.
# Change c[2] to 888.
# Check whether a changes.

# Q3: Create a sliced array d = a[1:4].
# Change d[0] to 777.
# Check whether a changes.

# Q4: Check memory sharing using:
# np.shares_memory(a, b)
# np.shares_memory(a, c)
```

```python
a = np.array([10, 20, 30, 40, 50])
b = a.copy()
b[1]=999
# not change because copy not effect the original array
print(a)
print(b)

# q2
c=a.view()
c[2]=888
# yes changes because view effect the original array also
print(a)
print(c)

# q3
d = a[1:4]
d[0]=777
# yes changes
print(a)
print(d)

# q4
np.shares_memory(a,b)
np.shares_memory(a,c)
```
