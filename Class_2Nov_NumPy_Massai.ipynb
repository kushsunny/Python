**Joining Arrays**

```python
a = np.array([10, 20, 30])
b = np.array([40, 50, 60])

jan = np.array([[100, 200], [300, 400]])
feb = np.array([[500, 600], [700, 800]])

print(np.concatenate((a,b)))
print(np.hstack((a,b)))
print(np.vstack((a,b)))
print(np.concatenate((jan,feb),axis=0))
print(np.concatenate((jan,feb),axis=1))
```
